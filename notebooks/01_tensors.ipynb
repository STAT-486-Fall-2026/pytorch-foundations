**Goals:** create tensors, understand shape and dtype, calculate matrix products, reshape data, and move between NumPy and PyTorch. Run each cell in order; try the exercises at the end. A tensor is a multidimensional array, much like a NumPy array, with support for automatic differentiation and devices.

```python
import numpy as np
import torch
torch.manual_seed(486)
x = torch.tensor([[1., 2., 3.], [4., 5., 6.]])
print(x, "\nshape:", x.shape, "dimensions:", x.ndim, "dtype:", x.dtype, "device:", x.device)
print("first row:", x[0], "last column:", x[:, -1])
```

## Operations and broadcasting
Elementwise multiplication `*` is different from matrix multiplication `@`. A length-three vector broadcasts across the rows of a `(2, 3)` matrix. Track the dimensions before calculating.

```python
offset = torch.tensor([10., 20., 30.])
print("broadcast addition:\n", x + offset)
print("elementwise square:\n", x * x)
weights = torch.tensor([[1., 0.], [0., 1.], [1., 1.]])  # (3, 2)
print("matrix product, (2, 3) @ (3, 2):\n", x @ weights)
print("column means:", x.mean(dim=0))
```

## Reshaping, types, and devices
`reshape` preserves the number and order of elements; it does not transpose axes. Floating point data is generally used for model inputs, while class indices usually use `torch.long`. NumPy conversion on a CPU can share memory; use `.clone()` or `.copy()` if you need an independent copy.

```python
print("reshape:", x.reshape(3, 2), "\ntranspose:", x.T)
print("add a batch dimension:", x[0].unsqueeze(0).shape)
arr = np.array([1, 2, 3], dtype=np.float32)
t = torch.from_numpy(arr)
print("NumPy -> tensor -> NumPy:", t, t.numpy())
print("integer labels:", torch.tensor([0, 1], dtype=torch.long))
device = torch.device("cpu")  # See setup guide for optional accelerators.
print("on device:", x.to(device).device)
```
